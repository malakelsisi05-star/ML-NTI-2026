```python
years = [str(year) for year in range(1980, 2014)]

years
```

```python
india = df[df['Country'] == 'India']
```

```python
plt.figure(figsize=(12,5))

plt.plot(years, india[years].values[0])

plt.title("Immigration from India to Canada (1980-2013)")
plt.xlabel("Year")
plt.ylabel("Number of Immigrants")

plt.xticks(rotation=45)

plt.show()
```

```python
top10 = df.sort_values(by='Total', ascending=False).head(10)

top10
```

```python
plt.figure(figsize=(10,6))

plt.bar(top10['Country'], top10['Total'])

plt.title("Top 10 Countries by Total Immigration")
plt.xlabel("Country")
plt.ylabel("Total Immigrants")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Total'], bins=20)

plt.title("Distribution of Total Immigrants")
plt.xlabel("Total Immigrants")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Total'], bins=20, kde=True)

plt.title("Distribution of Total Immigrants")
plt.xlabel("Total Immigrants")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.boxplot(df['Total'])

plt.title("Box Plot of Total Immigrants")
plt.ylabel("Total Immigrants")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.boxplot(y=df['Total'])

plt.title("Box Plot of Total Immigrants")
plt.ylabel("Total Immigrants")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df['1980'], df['2013'])

plt.title("Immigration in 1980 vs 2013")
plt.xlabel("1980")
plt.ylabel("2013")

plt.show()
```

```python
continent = df.groupby('Continent')['Total'].sum()

continent
```

```python
plt.figure(figsize=(8,8))

plt.pie(continent,
        labels=continent.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Immigration by Continent")

plt.show()
```

```python
plt.figure(figsize=(14,10))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
